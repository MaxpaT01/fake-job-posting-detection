# Fake Job Posting Detection - Machine Learning & NLP System

## 1. Project Overview & Problem Statement
Fraudulent job postings pose significant security and financial risks to job seekers worldwide. This project develops an end-to-end Machine Learning system capable of detecting scam job postings by analyzing unstructured textual data (`job description`, `requirements`, `company profile`) alongside structured metadata (`telecommuting`, `has_company_logo`, `has_questions`).

### Core Objectives:
1. Perform **Exploratory Data Analysis (EDA)** on the EMSCAD benchmark dataset (~18,000 postings).
2. Implement **Text Cleaning & Preprocessing** (HTML stripping, stopword removal, regex tokenization).
3. Extract numerical representations using **TF-IDF Vectorization** (Unigrams + Bigrams).
4. Combine text vectors with **scaled metadata signals** (missing company profile, missing logo).
5. Train and compare classification algorithms (**Logistic Regression**, **Multinomial Naive Bayes**, **Random Forest**, **Gradient Boosting**).
6. Evaluate performance using **Accuracy, Precision, Recall, F1-Score, and ROC-AUC**.
7. Deploy an interactive **Web Application & Dashboard** for real-time predictions.

## 6. Real-World Deployment & Limitations
The trained model is deployed as an interactive web app (`index.html` & `app.js`) running on GitHub Pages.
### Limitations:
1. **Adversarial Scams**: Fraud networks quickly adapt phrasing to evade keyword detectors.
2. **Early-stage Startups**: Legitimate startups without brand logos may be flagged as caution.
3. **Human-in-the-Loop**: High-risk postings should trigger human moderation rather than automated rejection.
